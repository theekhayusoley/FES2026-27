# 02 · Bucles `for`

Permeten executar un bloc de codi repetidament mentre **itera un iterable** (llista, cadena, `range`, ...).

## 📋 Iterar una llista

```python
fruites = ["poma", "pera", "mandarina", "plàtan"]
for fruita in fruites:
    print(fruita)
```

## 🔤 Iterar sobre qualsevol iterable

Una cadena de text també és iterable.

```python
cadena = "marc"
for caracter in cadena:
    print(caracter)
```

## 🔢 `enumerate()`

Dóna l'**índex** i el **valor** alhora.

```python
fruites = ["poma", "pera", "mandarina"]
for idx, valor in enumerate(fruites):
    print(f"L'índex és {idx} i la fruita és {valor}")
```

## 🪆 Bucles imbricats

```python
lletres = ["A", "B", "C"]
numeros = [1, 2, 3]

for lletra in lletres:
    for numero in numeros:
        print(f"{lletra}{numero}")
```

> 🔎 Per visualitzar l'execució pas a pas: [Python Tutor](http://pythontutor.com/visualize.html#mode=edit).

## 🛑 `break`

```python
animals = ["gos", "gat", "ratolí", "lloro", "peix", "canari"]
for idx, animal in enumerate(animals):
    print(animal)
    if animal == "lloro":
        print(f"El lloro està amagat a l'índex {idx}")
        break
```

## ⏭️ `continue`

```python
animals = ["gos", "gat", "ratolí", "lloro", "peix", "canari"]
for idx, animal in enumerate(animals):
    if animal == "lloro":
        continue
    print(animal)
```

## ✨ Comprensió de llistes (*list comprehension*)

Una forma compacta de crear llistes noves.

```python
animals = ["gos", "gat", "ratolí", "lloro", "peix", "canari"]
animals_majus = [animal.upper() for animal in animals]
print(animals_majus)
```

Mostra els números parells d'una llista:

```python
parells = [num for num in [1, 2, 3, 4, 5, 6] if num % 2 == 0]
print(parells)
```
