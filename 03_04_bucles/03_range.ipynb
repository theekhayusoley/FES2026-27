# 03 · `range()`

Permet crear una **seqüència de números**. És útil per a `for`, però no només per a això.

| Forma | Significat |
|---|---|
| `range(final)` | de 0 fins a `final - 1` |
| `range(inici, final)` | de `inici` fins a `final - 1` |
| `range(inici, final, pas)` | amb un increment (o decrement) de `pas` |

> ⚠️ El límit superior **no s'inclou**.

## Seqüència del 0 al 9

```python
for num in range(10):
    print(num)
```

## `range(inici, final)`

```python
for num in range(5, 10):
    print(num)
```

## `range(inici, final, pas)`

```python
for num in range(0, 100, 5):
    print(num)
```

## Números negatius i ordre invers

```python
for num in range(-5, 0):
    print(num)
```

```python
for num in range(10, 0, -1):
    print(num)
```

## Convertir un `range` en llista

```python
nums = range(1, 11)
llista_de_nums = list(nums)
print(llista_de_nums)
```

## Repetir alguna cosa N vegades

Si no necessitem la variable, per convenció fem servir `_`.

```python
for _ in range(5):
    print("fer cinc vegades alguna cosa")
```

Molt millor que fer-ho amb un `while`, on ens hem de recordar d'incrementar el comptador:

```python
num = 0
while num < 5:
    print("fer cinc vegades alguna cosa")
    num += 1
```
